```python
import plotly.express as px
import plotly.graph_objects as go
import plotly.subplots
from plotly.subplots import make_subplots
```

```python
merge = review_detail.merge(booking_detail, on=['booking_id', 'guest_id', 'hotel_id'], how='inner')
merge = merge.merge(guest_detail, on='guest_id', how='inner')
merge = merge.merge(hotel_detail, on='hotel_id', how='inner')
```

```python
### ROOM TYPE DISTRIBUTION
room_type = booking_detail['room_type'].value_counts().reset_index()
room_type.columns = ['room_type', 'count']
room_type

### ROOM REVENUE BY ROOM TYPE
room_revenue = booking_detail.groupby('room_type')['total_amount'].sum().reset_index()
room_revenue.columns = ['room_type', 'total_revenue']

#### BOOKING CHANNEL DISTRIBUTION
channels = booking_detail['booking_channel'].value_counts().reset_index()
channels

#### BOOKING CHANNEL REVENUE
channel_revenue = booking_detail.groupby('booking_channel')['total_amount'].sum().reset_index()
channel_revenue

fig = make_subplots(rows=2, cols=2, subplot_titles=('Room Type by Distribution', 'Room Revenue by Room Type','Channel Distribution','Total Revenue by Booking Channel'))
fig.add_trace(go.Bar(x=room_type['room_type'], y=room_type['count'], name='Room Type Distribution'), row=1, col=1)
fig.add_trace(go.Bar(x=room_revenue['room_type'], y=room_revenue['total_revenue'], name='Room Revenue by Room Type'), row=1, col=2)
fig.add_trace(go.Bar(x=channels['booking_channel'], y=channels['count'], name='Channel Distribution'), row=2, col=1)
fig.add_trace(go.Bar(x=channel_revenue['booking_channel'], y=channel_revenue['total_amount'], name='Channel Revenue'), row=2, col=2)

fig.update_layout(title_text='Booking Analysis', height=800, width=1000)
fig.show()
```
